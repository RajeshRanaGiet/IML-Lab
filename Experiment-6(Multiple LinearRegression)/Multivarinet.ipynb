```python
def transpose(matrix):
  rows = len(matrix)
  cols = len(matrix[0])
  t = []

  for j in range(cols):
    row = []
    for i in range(rows):
      row.append(matrix[i][j])
    t.append(row)

  return t

def multiply(A, B):
  result = []

  for i in range(len(A)):
    row = []
    for j in range(len(B[0])):
      sum_val = 0
      for k in range(len(B)):
        sum_val += A[i][k] * B[k][j]
      row.append(sum_val)
    result.append(row)
  return result

def inverse_2x2(M):
  det = M[0][0]*M[1][1] - M[0][1]*M[1][0]

  inv = [
      [M[1][1]/det, -M[0][1]/det],
      [-M[1][0]/det, -M[0][0]/det]
  ]
```
